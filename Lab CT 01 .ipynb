```python
# B

for name in st_dict:
  st_dict [name] = np.array (st_dict [name] )

print ( st_dict )
```

```python
# C

mark_avg = {}
for name in st_dict:
    total = 0
    cnt = 0
    for s in st_dict[name]:
        total += s
        cnt += 1
        mark_avg[name] = total / cnt

print ( "Avg : ", mark_avg)
print ( "Cnt : ", cnt)
```

```python
# A

numbers = [ 10, 25, 30, 45, 50, 60, 75, 80 ]

new_list = [ x for x in numbers if x%5==0 and x%10 != 0]
print ( "New list :",  new_list )
```

```python
# B

for i in range (len(new_list)) :
  if new_list [i] > 50:
    new_list[i] = new_list[i]/2

print ( new_list )

```

```python
# C

# new_list = [ 10, 25, 3,5, 30, 45, 50, 60, 75, 80, 2]

for i in range(len(new_list)):
    for j in range(len(new_list) - i - 1):
        if new_list[j] > new_list[j + 1]:
            new_list[j], new_list[j + 1] = new_list[j + 1], new_list[j]

print ("sorted list:", new_list)
```

```python
# D

mid = len(new_list) // 2
final = new_list[mid - 2: mid + 2]

print ("Extracted list:", new_list)
```

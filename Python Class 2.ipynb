````python
# array - homogeneous structure
# list - heterogeneous structure
lst = list()
print(type(lst))
lst = [10,20,30,40,60]
print(len(lst))
print(lst[0])
print(lst[-2])
````

````python
marks = []
marks.append(78)
print(marks)
````

````python
marks.append(80) # adds an element at the end of the list
print(marks)
marks.insert(1,'AI') # adds an element at the specified position
print(marks)
# marks.pop() # removes an element from the end of the list
marks.pop(1) # removes an element from the specified position
print(marks)
marks.remove(80) # removes an element from the list if it exists
print(marks.index(78))
````

````python
# searching
target = 80
if target in marks:
  marks.remove(target)
print(marks)
````

## List comprehension method

````python
squared = [i**2+3 for i in range(1,11) if i % 2 == 1]
print(squared)
````

````python
lst = [10,20,30,40,60,100,200]
#Iteration: index-based
for i in range(len(lst)):
  print(lst[i])

# Iteration: element-based
for i in lst:
  print(i)
````

# Dictionary

## Basics

````python
# Key-Value pairs
# Hashmap - (key,Value)
# Dictionary - Key:Value
# Keys must either be string or integer
s = 'our class will be held online from today' # (i,2), (o,3)
course = {
    'code':'CSE 3812',
    'name':'AI LAB',
    'credit hour':1,
    'duration':2.5,
    'grades':['A','A-','B'],
    827:'Class room',
    3.5: 'Avg'
}
print(type(course))
print(course)
````

## Indexing

````python
print(course['grades'])
print(course.get('name'))
````

## Iteration

````python
# course['student_grades']
print(course.keys()) # shows all the keys present in the dictionary
print(course.values()) # shows all the values present in the dictionary
print(course.items()) # shows all the key,value pairs in the dictionary
````

# Conditional Statement

```python
marks = int(input("Enter your marks: "))

if marks >= 90:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

# Dictionary

```python
student = {
    "name": "Jony",
    "age": 22,
    "dept": "CSE"
}

print(student["name"])
print(student["age"])

student["age"] = 23
student["cgpa"] = 3.75

print(student)
```

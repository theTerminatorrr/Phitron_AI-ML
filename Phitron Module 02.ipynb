**Structure of a function - Parameter**

```python
def welcome_message () :
  print ( "Hello User ")
```

```python
welcome_message()
```

```python
def welcome_message ( user ) :
  print ( f"Hello {user} ")
```

```python
welcome_message(" Jony ")
```

**Structure of a func - Return and Doc**

```python
def adder ( a,b ):
  sum = a+b
  return ( sum )
```

```python
m = adder ( 10, 13 ) * 5
print ( m )
```

```python
def adder ( a,b ):
  """
  add two numbers
  a -> int
  b -> int
  """

  sum = a+b
  return sum

m = adder ( 10, 13 ) * 5
print ( m )
```


**Positional Arguments**

```python
def adder ( a,b,c ):
  print ( f"{a},{b},{c}" )

adder ( 10, 30, 20)
```

```python
def adder ( *args ):
  print ( args )

adder ( 10, 30, 20)
```

```python
def adder ( *args ) :
  for i in args :
    print ( i )

adder ( 10, 20, 30, 70 )
```

**Keyword Arguments**

```python
def adder ( a,b,c ):
  print ( f"{a},{b},{c}" )

adder ( a=10, c=30, b=20)
```

```python
def adder (**kwargs):
  print ( kwargs )

adder ( a=10, c=30, b=20, d=40, n_estimator=50)
```

```python
def adder (**kwargs):
  for key, value in kwargs.items():
    print (f"key : {value}")
adder ( a=10, c=30, b=20, d=40, n_estimator=50)
```

**Return Type of a Function**

```python
def adder ( a,b ):
  sum = a+b
  return sum

print ( "Sum = ", adder( 10.5, 20 ))
print ( type( adder( 10.5, 20 )))
```

```python
def adder ( a,b ):
  return a, b

#unZip

x,y = adder ( 10.5, 20 )
print ( x )
print ( y )
```

**Iterator and Generator**

```python
s = { 10, 20, 30, 40, 50}
s_iter = iter(s)
print ( next (s_iter ) )
print ( next (s_iter ) )
print ( next (s_iter ) )
print ( next (s_iter ) )
print ( next (s_iter ) )

print ( "\n" )

for i in s :
  print ( i )
```

**Generator**

```python
list = [ x for x in range (500) ]
print ( list )

def data_loader (chunk_size, list) :
  for i in range ( 0, len(list) ,chunk_size ):
    yield list [ i : i+chunk_size ]

x = data_loader ( 5, list )
print  ( next (x) )
print  ( next (x) )
print  ( next (x) )
```

# Lambda , Map and Apply function


```python
import pandas as pd


df = pd.DataFrame({'temperature (F)': [32, 212, 98.6, 68]})
df

df['temperature (C)'] = df['temperature (F)'].apply(lambda x: (x - 32) * 5/9)
df
```

```python
import pandas as pd


df = pd.DataFrame({'A': [1, 2, 3, 4], 'B': [5, 6, 7, 8]})

df

df['A'] = df['A'].apply(lambda x: x ** 0.5)


df
```

#**File Handling Basics**

#File Read

```python
file = open ( "./sample_data/Sample.txt", "r")
content = file.read()

print ( content )
print ( type(content ))
```

```python
file = open ( "./sample_data/Sample.txt", "r")
content = file.readlines()

print ( content )
print ( type(content ))

file.close ()
print ( file.closed )
```

```python
with open ( "./sample_data/Sample.txt", "r") as file :
  content = file.readlines()
  print ( content )

print ( file.closed )
```

```python
with open ( "./sample_data/Sample.txt", "r") as file :
  for line in file :
    print ( line, end = "" )
```

```python
# Model Train

try :
  file = open ( "./sample_data/data1.txt", "r" )
  file.read()

except Exception as e :
  print ( e )
```

```python
# Model Train

try :
  file = open ( "./sample_data/data1.txt", "r" )
  file.read()

except Exception as e :
  print ( e )
else :
  #model train
  print (file.read())
```

```python
# model train


try:
  file = open("./sample_data/data1.txt","r")
except Exception as e:
  print(e)
else:
  # model train
  print(file.read())
finally:
  print("Gpu is stopped")
```
